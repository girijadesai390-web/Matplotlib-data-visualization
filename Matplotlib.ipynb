```python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
study_hours=np.random.uniform(1,10,size=50)
exam_scores=40+(study_hours*5.5)+np.random.normal(0,5,size=50)
plt.scatter(study_hours,exam_scores,color="red")
plt.xlabel="Study Hours"
plt.ylabel="Exam Scores"
plt.title="Scatter Plot"
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
customer_ages = np.random.normal(loc=35, scale=8, size=1000)
plt.figure(figsize=(8,5))
plt.hist(customer_ages, bins=25, color='#0ea5e9', edgecolor='white', alpha=0.8)
plt.title('Customer Age Distribution', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Age Group', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed (42)
class_A = np.random.normal(75, 10, 100)
class_B = np.random.normal(82, 7, 100)
class_C = np.random.normal(68, 15, 100)
data_to_plot = [class_A, class_B, class_C]
plt.figure(figsize=(8, 5))
plt.boxplot(
data_to_plot,
labels=['Class A', 'Class B', 'Class C'],
patch_artist=True
)
plt.title('Exam Score Distribution Across Classes', fontsize=14, fontweight='bold', pad=15)
plt.ylabel('Exam Scores', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
correlation_matrix = np.random.uniform(0.1, 1.0, size=(5,5))
np.fill_diagonal (correlation_matrix, 1.0)
features = ['Feature 1', 'Feature 2', 'Feature 3', 'Feature 4', 'Feature 5']
plt.figure(figsize=(7, 6))
plt.imshow(correlation_matrix, cmap='magma', interpolation='nearest')
for i in range(5):
  for j in range(5):
     plt.text(j, i, f"{correlation_matrix[i, j]:.2f}",
ha="center", va="center", color="white" if correlation_matrix[i, j] < 0.6
else "black")
plt.xticks(np.arange(5), features, rotation=45)
plt.yticks(np.arange(5), features)
plt.title('Feature Correlation Heat Map Matrix', fontsize=14, fontweight='bold', pad=15)
plt.colorbar (label='Correlation Strength Scale')
plt.tight_layout()
plt.show()
```
